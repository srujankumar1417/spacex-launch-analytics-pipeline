```python
import requests
import pandas as pd
import datetime as dt
import json
```

```python
def create_json(resourse):
    url = f'https://api.spacexdata.com/v4/{resourse}'
    timestamp_launch = dt.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    path_json = f'{resourse}.json'  
    end_point = resourse
   
    try:
        response = requests.get(url)
        response.raise_for_status() 
        data = response.json()
        if not data:
            print(f"No data found for resource: {resourse}")
            return None
        bronze_data = {
            "source_url": url,
            "date_time": timestamp_launch,
            "status_code": response.status_code,
            "record_count": len(data),
            "source_data": data,
            "end point " : end_point
        }
        with open(path_json,"w") as file:
            json.dump(bronze_data , file , indent  = 4 )
            print(f'{resourse}.json file successfully created')
            return 

    except requests.RequestException as e:
        print(f"Error fetching SpaceX data: {e}")
        return None
    

```

```python
create_json('launches')
```

```python
create_json('rockets')
```

```python
create_json('launchpads')
```
